```python
df = pd.read_csv(
    file_path,
    sep="\t",
    header=None,
    names=["datetime", "open", "high", "low", "close", "volume"]
)

# Convert datetime column into actual datetime objects
df["datetime"] = pd.to_datetime(df["datetime"])

# Convert price and volume columns into numeric values
numeric_columns = ["open", "high", "low", "close", "volume"]

df[numeric_columns] = df[numeric_columns].apply(
    pd.to_numeric,
    errors="coerce"
)

print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("First candle:", df["datetime"].min())
print("Last candle:", df["datetime"].max())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate timestamps:")
print(df["datetime"].duplicated().sum())
```

```python
# Calculate the time difference between consecutive candles
time_diff = df["datetime"].diff()

print("Most common time intervals:")
display(time_diff.value_counts().head(15))

print("\nIs the dataset chronologically ordered?")
print(df["datetime"].is_monotonic_increasing)

print("\nNumber of intervals longer than 5 minutes:")
print((time_diff > pd.Timedelta(minutes=5)).sum())

print("\nLargest time gap:")
print(time_diff.max())
```
